```python
import pandas as pd
df=pd.read_csv(r'C:\Users\saido\Desktop\ApplicationGestionAi\CreationModelAi\ModelAi\livres_nettoyes.csv')
df
```

```python
df.drop(columns=['availability'],inplace=True)
df

```

```python
df['stock'] = 10
df['rating'] = df['rating'].astype(int)

# تحويل الأعمدة لنصوص (String)
df['description'] = df['description'].astype(str)
df['image_url'] = df['image_url'].astype(str)
```

```python
df['id_livre'] = range(1, len(df)+1)

# ترتيب الأعمدة بحيث يصبح id_livre هو الأول
cols = ['id_livre'] + [col for col in df.columns if col != 'id_livre']
df = df[cols]
df
```

```python
# إعادة تسمية الأعمدة في DataFrame
df = df.rename(columns={
    'title': 'titre',
})
df.columns
```

```python
from sqlalchemy import create_engine, Integer, String, DECIMAL

engine = create_engine('mysql+mysqlconnector://root:@localhost:3306/gestionbeb')

df.to_sql(
    name='livres',
    con=engine,
    if_exists='append',
    index=False,
    dtype={
        'titre': String(255),
        'description': String(255),
        'image_url': String(255),
        'stock': Integer,
        'rating': Integer,
        'price': DECIMAL(10,2)  # باش يكون رقم عشري
    }
)
```
